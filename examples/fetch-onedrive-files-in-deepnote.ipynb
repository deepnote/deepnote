## Fetch a file from OneDrive

OneDrive allows us to fetch publicly shared files by taking a share URL and passing it to the [`shares_get` OneDrive API](https://docs.microsoft.com/en-us/onedrive/developer/rest-api/api/shares_get). The function `get_onedrive_download` creates the URL to download any shared file.

```
# Convert a share link to a direct download
# See https://docs.microsoft.com/en-us/onedrive/developer/rest-api/api/shares_get
def get_onedrive_download(share_link):
    base64_link_bytes = base64.b64encode(bytes(share_link, 'utf-8'))
    base64_link_string = base64_link_bytes.decode('utf-8').replace('/','_').replace('+','-').rstrip("=")
    return f"https://api.onedrive.com/v1.0/shares/u!{base64_link_string}/root/content"

ONEDRIVE_SHARE_LINK = "https://1drv.ms/x/s!AjP7y6eiYDDSafV7BusEaDHrI0o?e=cL0Fqi"

get_onedrive_download(ONEDRIVE_SHARE_LINK)
```

We then use pandas to read the URL into a dataframe directly. Deepnote summarizes and paginates dataframes automatically. You can filter and sort each column to drill into the data.

```
df = pd.read_excel(get_onedrive_download(ONEDRIVE_SHARE_LINK))
df
```
