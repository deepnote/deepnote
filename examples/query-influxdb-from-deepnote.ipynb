## Query data from InfluxDB

We use our API key to compose a query in the Flux query language and run it against our InfluxDB instance.

The Python client returns the result as a Pandas dataframe, which means we can leverage Deenote's built-in dataframe analysis. You can filter and sort each column to drill into the data.

```
query_api = client.query_api()

query = """import "influxdata/influxdb/sample"

sample.data(set: "airSensor")
  |> filter(fn: (r) => r._field == "temperature")
  |> range(start: -45m)
  |> aggregateWindow(every: 1m, fn: mean)"""

# influxdb_client gives us Pandas dataframes out-of-the-box
df = query_api.query_data_frame(query)

df
```
